```python
import networkx as nx
import matplotlib.pyplot as plt

%matplotlib inline

my_edges = [
    ('A', 'B', 4), ('A', 'C', 7), ('A', 'D', 1), 
    ('B', 'C', 5), ('B', 'D', 4), ('C', 'D', 3), 
    ('C', 'E', 2), ('D', 'E', 2), ('C', 'F', 3), ('E', 'F', 6)
]

G = nx.Graph()
G.add_weighted_edges_from(my_edges)

# SỬ DỤNG KAMADA_KAWAI_LAYOUT: Giúp các đỉnh giãn ra xa nhau, hạn chế cắt cạnh
pos = nx.kamada_kawai_layout(G)

plt.figure(figsize=(10, 8))

# Vẽ đỉnh và cạnh
nx.draw(
    G, pos, 
    with_labels=True, 
    node_color='#87CEEB', # Màu xanh da trời nhạt (SkyBlue)
    node_size=1000, 
    font_size=15, 
    font_weight='bold',
    edge_color='#555555', # Màu xám đậm
    width=2.5
)

# Vẽ trọng số
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(
    G, pos, 
    edge_labels=edge_labels, 
    font_color='red', 
    font_size=13,
    font_weight='bold',
    rotate=False,  # BÍ QUYẾT: Không xoay chữ theo cạnh
    bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.8) # Thêm nền trắng cho chữ dễ đọc
)

plt.title("Đồ thị có trọng số bài tập", fontsize=18, fontweight='bold', color='navy')
plt.axis('off') # Ẩn trục tọa độ x,y cho giống hình vẽ toán học
plt.show()
```
